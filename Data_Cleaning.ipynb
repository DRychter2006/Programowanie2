#Data cleaning
- 1 i 2 -> 0 i 1
- Płeć na liczby M -> 0 oraz K -> 1
- Zamiana zmiennej docelowej z YES/NO na 1/0

```python
# Usunięcie spacji na początku i na końcu nazwy kolumny
df.columns = df.columns.str.strip()

# Kolumny z wartościami 1/2
columns_to_fix = [
    'SMOKING',
    'YELLOW_FINGERS',
    'ANXIETY',
    'PEER_PRESSURE',
    'CHRONIC DISEASE',
    'FATIGUE',
    'ALLERGY',
    'WHEEZING',
    'ALCOHOL CONSUMING',
    'COUGHING',
    'SHORTNESS OF BREATH',
    'SWALLOWING DIFFICULTY',
    'CHEST PAIN'
]

# Zamiana 1 -> 0, 2 -> 1
for col in columns_to_fix:
    if col in df.columns:
        df[col] = df[col].replace({1: 0, 2: 1})

# Zamiana płci na liczby
df['GENDER'] = df['GENDER'].map({
    'M': 0,
    'F': 1
})

# Zamiana zmiennej docelowej(Lung Cancer)
df['LUNG_CANCER'] = df['LUNG_CANCER'].map({
    'NO': 0,
    'YES': 1
})

# Podgląd
print(df.head())

# Zapis nowego pliku
new_path = "/content/drive/MyDrive/Programowanie2/dataset_clean.csv"
df.to_csv(new_path, index=False)

print(f"\nZapisano plik:\n{new_path}")
```
